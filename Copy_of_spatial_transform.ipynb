```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

#Baca Citra
img = cv2.imread('harley.jpg')

cv2_imshow(img)

#Tampilkan Citra
# plt.imshow(img)
# plt.show()
```

```python
#translasi citra

M_trans = np.float32([[1,0,100],[0,1,50]])
translated = cv2.warpAffine(img, M_trans, (img.shape[1], img.shape[0]))

cv2_imshow(translated)
```

```python
M_rot = cv2.getRotationMatrix2D((img.shape[1]/2, img.shape[0]/2), 45, 1)
rotated = cv2.warpAffine(img, M_rot, (img.shape[1], img.shape[0]))

cv2_imshow(rotated)
```

```python
#scaled

scaled = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)

cv2_imshow(scaled)
```

```python
#tampilkan semua


plt.figure(figsize=(10,6))
plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(translated, cv2.COLOR_BGR2RGB))
plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(scaled, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
#interplorasi

# Scaling dengan berbagai metode interpolasi
nearest = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_NEAREST)
bilinear = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)
bicubic = cv2.resize(img, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)

plt.figure(figsize=(10,6))
plt.subplot(2,2,1)
plt.imshow(nearest)
plt.subplot(2,2,2)
plt.imshow(bilinear)
plt.subplot(2,2,3)
plt.imshow(bicubic)
plt.show()
```
